```python
import pandas as pd
import numpy as np

casas = pd.read_csv("houses_sp.csv", sep=",")

# 1) há valores nulos que possam comprometer a análise dos dados?

casas.isnull().sum() * 100/casas.shape[0]

```

```python
# Por se tratar de uma base de dados de imóveis para aluguel, levando em consideração o tamanho da base de dados e o percentual de itens com dados nulos, 
# vamos excluir as linhas com valores nulos nas colunas de area e quartos.

casas_filtrada = casas[casas['area'].notnull() & casas['rooms'].notnull()]

casas_filtrada.isnull().sum() * 100/casas_filtrada.shape[0]
```

```python
# a presença de vagas de estacionamento pode ser fator determinante para o valor do aluguel, de forma que vamos transformar os valores nulos da coluna correspondente em 0.

casas_filtrada['parking spaces'] = casas_filtrada['parking spaces'].fillna(0)

casas_filtrada.isnull().sum() * 100/casas_filtrada.shape[0]
```

```python
# 2) qual a relação entre o número de vagas de estacionamento e a média do aluguel?

casas_filtrada.groupby('parking spaces').agg({'rent': 'mean'})

```
