```python
import pandas as pd
import numpy as np

casas = pd.read_csv("houses_sp.csv", sep=",")
```

1) há valores nulos que possam comprometer a análise dos dados?


```python
casas.isnull().sum() * 100/casas.shape[0]
```

! Por se tratar de uma base de dados de imóveis para aluguel, levando em consideração o tamanho da base de dados e o percentual de itens com dados nulos, 
vamos excluir as linhas com valores nulos nas colunas de area e quartos.

```python
casas_filtrada = casas[casas["area"].notnull() & casas["rooms"].notnull()]

casas_filtrada.isnull().sum() * 100/casas_filtrada.shape[0]
```

A presença de vagas de estacionamento pode ser fator determinante para o valor do aluguel, de forma que vamos transformar os valores nulos da coluna correspondente em 0, 
indicando ausência de vagas.

```python
casas_filtrada["parking spaces"] = casas_filtrada["parking spaces"].fillna(0)

casas_filtrada.isnull().sum() * 100/casas_filtrada.shape[0]
```

2) qual a relação entre o número de vagas de estacionamento e a média do total do aluguel, incluindo taxas, medida com base na mediana?

```python
casas_filtrada.groupby("parking spaces").agg({"total": "median"})
```

! verifica-se uma correspondência direta entre as vagas de estacionamento e o aluguel, salvo casos pontuais, os quais precisam de análise mais detalhada.

3) porque imóveis com menos vagas de estacionamento podem apresentar valor de aluguel maior do que imóveis com mais vagas em algumas situações?

! Observando os dados dos imoveis com 4, 5, 7, 8, e 12 vagas de estacionamento, consideramos que um fator que pesa para o valor total do aluguel é a taxa de condomínio,
representada pela coluna "hoa".

4) Como o valor do aluguel varia de acordo com o tipo de imóvel?

```python
casas_filtrada.groupby(["type"])[["rent","total","hoa"]].describe()
```

! Verificamos que o valor do aluguel de casas tende a ser maior que o valor para apartamentos, contudo o valor total para apartamentos se torna maior em razão principalmente da taxa de condomínio (coluna "hoa").
! chama atenção o valor máximo da taxa de condomínio observada nesta anílise (220000), a qual é maior inclusive que o valor do aluguel, provavelmente indicando um erro na base de dados.

5) Quantos imóveis apresentam a taxa de condomínio maior que o valor do aluguel?

```python
casas_filtrada[casas_filtrada["hoa"] > casas_filtrada["rent"]].sort_values("hoa", ascending=False)
```

! No total, 142 imóveis apresentam valor de taxa de condomínio maior que o valor do aluguel. Essas observações possivelmente representam erros na base de dados.

6) Há impacto significativo dos imóveis cujo valor do condomínio é maior que o valor do aluguel na análise global?

```python
casas_filtrada[casas_filtrada["hoa"] > casas_filtrada["rent"]].value_counts().sum()/casas_filtrada.shape[0]
```

! Um impacto de 2% numa base de dados de 7000 linhas parece representar um pequeno impacto na análise global.

7) Como se distribui o impacto dos imóveis com valor de condomínio maior que o valor de aluguel baseado no número de quartos? seria razoável excluir esses imoveis da base de dados?

```python
casas_filtrada.groupby(["rooms"])[["rent", "hoa"]].agg(["size", "mean", "median",  "min", "max"])
```

```python
casas_filtrada[(casas_filtrada["rooms"] >= 5) & (casas_filtrada["rooms"] <= 6)].value_counts().sum()/casas_filtrada.shape[0]
```

! Tomando por base os valores das médias de medianas para os valores de aluguel e taxa de condomínio, verificamos que o impacto na análise se concentra nos imóveis com 5 e 6 quartos. 
Novamente, temos um impacto da ordem de 2% da base de dados, o que pode ser considerado um impacto pouco significativo para a análise global. No entanto, caso se considere analisar recortes por número de quartos, a exclusão dessas observações deve ser considerada.
